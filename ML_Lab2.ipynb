Section 1: Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('data.csv')

df = df.dropna(subset=['categorical_1', 'numerical_1', 'numerical_2'])

plt.figure(figsize=(10, 6))
plt.plot(df['categorical_1'], df['numerical_1'], label='Shirts', color='blue', linewidth=2, marker='o')
plt.plot(df['categorical_1'], df['numerical_2'], label='Trousers', color='green', linewidth=2, marker='o')
plt.plot(df['categorical_1'], df['numerical_1'] * 0.8, label='Shoes', color='red', linewidth=2, marker='o')

plt.xlabel('Month (categorical_1)')
plt.ylabel('Units Sold')
plt.title('Monthly Sales: Shirts, Trousers, and Shoes')
plt.legend(loc='upper right')
plt.show()

plt.figure(figsize=(10, 6))
plt.scatter(df['categorical_1'], df['numerical_2'], label='Trousers', color='purple', s=100)
plt.xlabel('Month (categorical_1)')
plt.ylabel('Trousers Sold')
plt.title('Monthly Sales for Trousers')
plt.legend(loc='upper right')
plt.grid(True, linewidth=3, linestyle='--')
plt.show()

plt.figure(figsize=(10, 6))
x = np.arange(len(df['categorical_1']))
width = 0.35

plt.bar(x - width/2, df['numerical_1'], width, label='T-Shirts', color='cyan')
plt.bar(x + width/2, df['numerical_2'], width, label='Shoes', color='orange')

plt.xticks(x, df['categorical_1'])
plt.xlabel('Month (categorical_1)')
plt.ylabel('Units Sold')
plt.title('Monthly Sales: T-Shirts vs Shoes')
plt.legend(loc='upper left')
plt.grid(True, linewidth=2, linestyle='--')


plt.savefig('tshirts_shoes_sales.jpeg', format='jpeg', dpi=400)
plt.show()
```
