Section 1 - General Data Processing

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv('data.csv')
print("Original DataFrame (first 3 rows):\n", df.head(3), "\n")

print("Q1 - Missing values per column:\n", df.isnull().sum(), "\n")

cols_with_missing = df.columns[df.isnull().any()].tolist()
print("Columns that have missing values:", cols_with_missing, "\n")

df_one_hot = pd.get_dummies(df, columns=['categorical_1', 'categorical_2'])
print("Q2 - One-Hot Encoded DataFrame (first 3 rows):\n", df_one_hot.head(3), "\n")

df_label_encoded = df.copy()
le = LabelEncoder()
df_label_encoded['categorical_1'] = le.fit_transform(df_label_encoded['categorical_1'].astype(str))
df_label_encoded['categorical_2'] = le.fit_transform(df_label_encoded['categorical_2'].astype(str))
print("Q3 - Label Encoded DataFrame (first 3 rows):\n", df_label_encoded[['categorical_1', 'categorical_2']].head(3), "\n")

df_standardized = df.copy()
scaler = StandardScaler()
df_standardized[['numerical_1', 'numerical_2']] = scaler.fit_transform(df_standardized[['numerical_1', 'numerical_2']])
print("Q4 - Standardized numerical columns (first 3 rows):\n", df_standardized[['numerical_1', 'numerical_2']].head(3))
```

2. Section 2 - Sentiment Analysis

```python
import pandas as pd
import string
import nltk
from nltk.tokenize import word_tokenize
import warnings
warnings.filterwarnings("ignore")

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

sentiment_df = pd.read_csv('sentiment_analysis.csv')
print("Q1 - DataFrame Loaded (first 3 rows):\n", sentiment_df.head(3), "\n")

print("Missing values before fill:\n", sentiment_df.isnull().sum(), "\n")
sentiment_df = sentiment_df.fillna('Unknown')
print("Q2 - Missing values after fill:\n", sentiment_df.isnull().sum(), "\n")

def remove_punct(text):
    if isinstance(text, str):
        return text.translate(str.maketrans('', '', string.punctuation))
    return text

sentiment_df['text_clean'] = sentiment_df['text'].apply(remove_punct)
print("Q3 - Text without punctuation (first 3 rows):\n", sentiment_df[['text', 'text_clean']].head(3), "\n")

sentiment_df['text_tokenized'] = sentiment_df['text_clean'].apply(lambda x: word_tokenize(x) if isinstance(x, str) else x)
print("Q4 - Tokenized Text (first 3 rows):\n", sentiment_df[['text_clean', 'text_tokenized']].head(3), "\n")

sentiment_df['text_rejoined'] = sentiment_df['text_tokenized'].apply(lambda x: ' '.join(x) if isinstance(x, list) else x)
print("Q5 - Rejoined Text (first 3 rows):\n", sentiment_df[['text_tokenized', 'text_rejoined']].head(3))
```
